for columna in columnas_categoricas:

```python
for columna in columnas_categoricas:
    moda = df[columna].mode()[0]
    df[columna] = df[columna].fillna(moda)

print("Valores categoricos corregidos")
```

Se importan matplotlib.pyplot y seaborn para generación de gráficos.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

df["country"].value_counts(): Cuenta cuántas veces aparece cada país en la tabla

.head(10): Toma únicamente los primeros 10 países de esa lista, es decir, el Top 10 de los que más tienen.

.reset_index(): Convierte esa información en una tabla limpia. Sin esto, los nombres de los países se quedan como "etiquetas" en lugar de ser una columna normal.

"country": Será el nombre de la columna que contiene los países.

"count": Será el nombre de la columna que contiene la cantidad total de títulos de cada país.

```python
top_paises = df["country"].value_counts().head(10).reset_index()
top_paises.columns = ["country", "count"]
```

data=top_paises: Le dice a la función de dónde sacar los datos. x="count": Pone en el eje horizontal (abajo) la cantidad de películas y series. y="country": Pone en el eje vertical (a la izquierda) los nombres de los países. Como los textos van en el eje $Y$, la gráfica se dibuja de forma horizontal.

plt.title("Top 10 Países con más Contenido") Le pone el título principal a la gráfica en la parte superior.

plt.xlabel("Cantidad de peliculas y series")Le pone una etiqueta o texto al eje horizontal ($X$) para indicar qué significan los números que se ven abajo.

plt.ylabel("País")Le pone una etiqueta al eje vertical ($Y$) para indicar que ahí están los nombres de los países.

plt.show() Muestra la gráfica final en la pantalla de tu Jupyter Notebook para que la puedas ver.

```python
sns.barplot(data=top_paises, x="count", y="country")
plt.title("Top 10 Países con más Contenido")
plt.xlabel("Cantidad de peliculas y series")
plt.ylabel("País")
plt.show()
```

df["type"] == "Movie"
Compara la columna "type" de la tabla para ver en qué filas el valor es exactamente "Movie". Esto devuelve una lista de respuestas con True (verdadero) o False (falso) para cada registro.

df[...]
Utiliza el resultado de la comparación para filtrar la tabla original (df). Solo se selecciona y conserva aquellas filas donde el tipo resultó ser "Movie", dejando fuera todas las series de televisión (TV Shows).

.copy() Crea una copia independiente en la memoria de la computadora con esta nueva información filtrada.

```python
df_movies = df[df["type"] == "Movie"].copy()
```

```python
df_movies["duration_min"] = pd.to_numeric(
    df_movies["duration"].astype(str).str.replace(" min", ""), 
    errors="coerce"
)
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x="release_year",
    y="duration_min",
    data=df_movies
)
plt.title("Año de Estreno vs Duración de Películas (en Minutos)")
plt.xlabel("Año de Lanzamiento")
plt.ylabel("Duración (minutos)")
plt.show()
```

```python
import pandas as pd
import plotly.express as px
```

```python
df_paises = df.dropna(subset=["country"]).copy()
df_paises = df_paises.assign(
    country=df_paises["country"].str.split(r"\s*,\s*")
).explode("country")
```

```python
df_paises["country"] = df_paises["country"].replace({
    "USA": "United States",
    "United States of America": "United States",
    "UK": "United Kingdom"
})
```

```python
df_paises = df_paises.groupby("country").size().reset_index(name="count")

# Crear mapa usando el número REAL, pero topando la escala de color en 500 para máximo contraste
fig = px.choropleth(
    df_paises,
    locations="country",
    locationmode="country names",
    color="count",  # Usamos la cantidad real directamente
    range_color=[0, 500],  # Todos los países con 500+ títulos (EE. UU., India, etc.) tendrán el rojo más intenso
    hover_name="country",
    hover_data={"count": ":,"},  # Muestra el número total real al pasar el ratón
    color_continuous_scale="Reds",
    title="Distribución Mundial de Producción de Contenido en Netflix"
)
```

```python
fig.update_layout(
    margin={"r": 0, "t": 50, "l": 0, "b": 0},
    geo=dict(
        showframe=False,
        showcoastlines=False,
        projection_type="natural earth"
    ),
    coloraxis_colorbar=dict(title="Títulos")
)

fig.show()
```
